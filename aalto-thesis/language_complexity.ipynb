# Language complexity
To evaluate the langauge complexity in the survey, I can utilize the [Flesch-Kincaid readability test](https://en.wikipedia.org/wiki/Flesch%E2%80%93Kincaid_readability_tests). In addition, I can calculate character, word, and sentence counts and a rough duration for how much time it takes to read a certain piece of text.

I'm using the `textstat` package developed by Shivam Bansal and Chaitanya Aggarwal. `textstat` is an [open source package](https://github.com/textstat/textstat/tree/main?tab=readme-ov-file) "to calculate statistics from text. It helps to determine readability, complexity, and grade level."

```python
import textstat
import pandas as pd

df = pd.read_csv('/content/all_text.csv')
display(df.head())
print(df['string'][0])
```

```python
for i in range(len(df)):
  text = df['string'][i]
  ease = textstat.flesch_reading_ease(text) # This doesn't work in this context. The text is not one text meant to be read as one piece.
  read_time = textstat.reading_time(text, ms_per_char=53.65) # In seconds. I cannot use both the read_time and word_count. Problem with collinearity.
  word_count = textstat.lexicon_count(text, removepunct=True)
  char_count = textstat.char_count(text, ignore_spaces=True)

  print(word_count)
```
